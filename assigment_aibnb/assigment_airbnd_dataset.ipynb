# Detect outliners

```python
# See the old file and see the how to make a outliiner

import numpy as np
def detect_outlier(col):
    Q1 = np.percentile(col,25)
    Q3 = np.percentile(col,75)
    IQR = Q3 - Q1
    
    lower_boundr = Q1 - 1.5 * IQR
    upper_boundr = Q3 + 1.5 * IQR
    return lower_boundr,upper_boundr
```

```python
lb,up =  detect_outlier(data.calculated_host_listings_count)
```
