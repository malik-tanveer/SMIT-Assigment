```python
import pandas as pd
data = pd.read_csv("Student_Performance.csv")
```

```python
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

```python
data.columns
```

```python
data.columns = [
    'Hours_Studied', 'Previous_Scores', 'Extracurricular_Activities',
       'SleepHours', 'Question_Papers_Practiced', 'Performance_Index'
]
```

```python
data.head()
```

```python
data.Extracurricular_Activities.unique()
```

```python
data.Extracurricular_Activities = data.Extracurricular_Activities.map({
    "Yes" : 1,
    "No" : 0
})
```

```python
X = data.drop("Performance_Index",axis=1)
y = data[["Performance_Index"]]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=35
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
student_model = LinearRegression()
```

```python
student_model.fit(X_train, y_train)
```

```python
y_pred = student_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
def student_model_predict():
    hours_study = int(input("Enter a Hours Study:"))
    previous_scroe = int(input("Enter a your previous score:"))
    activiites = int(input("Enter a Extra Circular Activites:"))
    sleep_hours = int(input("Enter a your sleep hours:"))
    question_paper = int(input("Enter a your question paper practiced:"))

    new_data = pd.DataFrame({
        "Hours_Studied" : [hours_study],
        "Previous_Scores" : [previous_scroe],
        "Extracurricular_Activities" : [activiites],
        "SleepHours": [sleep_hours],
        "Question_Papers_Practiced" : [question_paper]
    })

    prediction = student_model.predict(new_data)

    print(prediction[0])
```
