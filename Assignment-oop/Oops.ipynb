1. Create a class 'BankAccount' with attributes account holder name, account number, and
balance. Add methods for deposit and withdrawal. Implement logic to prevent withdrawal if balance
is insufficient.

```python
class Bankaccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposit")

    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def balance_chck(self):
        print(f"""Total Balance in Your Account {self.balance}""")
```

3. Create a class 'ElectricityBill' for a KE-electric consumer. Include units consumed and calculate
bill based on slab rates (e.g., different price per unit for different ranges).

```python
class ElectricityBill:
    def __init__(self, name, units):
        self.name = name
        self.units = units

    def cal_bill(self):
        units = self.units
        bill = 0 # bill inital value is 0
        
        if units <= 100:
            bill = units * 5
        elif units <= 300:
            bill = 100 * 5 + (units - 100) * 8
        elif units <= 700:
            bill = 100 * 5 + 200 * 8 + (units - 300) * 12
        else:
            bill = 100 * 5 + 200 * 8 + 400 * 12 + (units - 700) * 18
        return bill

    def display_bill(self):
        print("Units Consumed:", self.units)
        print("Total Bill: Rs.", self.cal_bill())
```

6. Design an abstract class 'Shape' with a method 'area()'. Create subclasses 'Circle' and
'Rectangle' and implement area calculation logic.

```python
from abc import ABC, abstractmethod
import math

# hum Shapes ka object nahi bnya ga hum is ka chidlern ki class ka obj bnya ga 
class Shapes(ABC):

    @abstractmethod
    def area(self): # rule defined ka area class har jga hogi
        pass

class Circle(Shapes): # circle class
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2  # math.pi PI ki value print krta hai 

class Rectangle(Shapes): # rectangle class
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
```

```python
circle = Circle(5)
rectangle = Rectangle(4, 6)
```

```python
print("Circle Area:", circle.area())
print("Rectangle Area:", rectangle.area())
```

```python
class libraray:
    def __init__(self):
        self.books = ["Python", "Java", "C++", "C", "HTML", "CSS","JS"]

    def borow_book(self, book):
        if book in self.books:
            self.books.remove(book)
            print(book, "issued success")
        else:
            print(book, "not available")

    def return_book(self, book):
        self.books.append(book)
        print(book, "returned success")
```

## 11.Create a Python program using OOP Design an airline reservation system.

### *Requirements:*
- Create abstract class Flight
- Subclasses: Economy, Business, FirstClass
- Use encapsulation for ticket price
- Apply discount (children & senior citizens)
- Add seat availability system
- Create Passenger booking system
- Add payment + installment option
- Apply airport tax (12%)
- Generate ticket with unique Booking ID
- Save booking data in file

```python
from abc import ABC, abstractmethod
import random # for num genrate

# Parent class
class flight(ABC):
    def __init__(self, price, seats):
        self._price = price
        self.seats = seats

    def get_price(self):
        return self._price

    def set_price(self, price):
        self._price = price

    @abstractmethod
    def calculate_price(self): # sirf is class ka lia rule bnya hia is ko use krna hai child ma
        pass

# child Economy 
class Economy(flight):
    def calculate_price(self):
        return self.get_price()

# child Busines
class Business(flight):
    def calculate_price(self):
        return self.get_price()

# Child firstclass
class FirstClass(flight):
    def calculate_price(self):
        return self.get_price() 


# Passenger Booking System
class Passenger:
    def __init__(self, name, age, flight):
        self.name = name
        self.age = age
        self.flight = flight
        self.booking_id = random.randint(0,1000) # random genrate

    # age ka opar discount mila ga
    def apply_discount(self, price):
        if self.age < 12:
            return price * 0.5
        elif self.age > 60:
            return price * 0.7
        else:
            return price

    # final ticket book kra ga
    def book_ticket(self):
        if self.flight.seats <= 0:
            return "seat is not avaliable"

        # Calculate original ticket price based on flight class
        base_price = self.flight.calculate_price()
        discounted_price = self.apply_discount(base_price) # Apply age-based discount
        discount_amount = base_price - discounted_price  # Calculate total discount amount
        tax_amount = discounted_price * 0.12 # Calculate 12% tax on discounted price
        final_price = discounted_price + tax_amount # Add tax to get final payable amount
        self.flight.seats -= 1 # Reduce available seats after successful booking

        # file bna kar is ma detail save kr do 
        with open("bookings.txt", "a") as f:
            f.write(
                f"Booking ID: {self.booking_id}, "
                f"Name: {self.name}, "
                f"Base Price: {base_price}, "
                f"Discount: {discount_amount}, "
                f"After Discount: {discounted_price}, "
                f"Tax: {tax_amount}, "
                f"Final Price: {final_price}\n"
            )

        return final_price
```

## 11.Create a Python program using OOP concepts for a shop that sells AC, Refrigerator, and Washing
Machine. Your program must follow real shop flow where customer selects products and system
generates invoice.
Requirements:
1. Create abstract class Product
2. Create subclasses: AC, Refrigerator, WashingMachine
3. Use encapsulation for price
4. Apply discount based on conditions
5. Add stock system (no purchase if stock is 0)
6. Create Customer class with cart
7. Add Payment and Installment system
8. Add 18% tax
9. Generate Invoice with UUID
10. Save record in text file

```python
from abc import ABC, abstractmethod
import uuid # unique id gen

# parent class
class Product(ABC):
    def __init__(self, name, price, stock):
        self.name = name
        self._price = price
        self.stock = stock

    def get_price(self):
        return self._price

    @abstractmethod
    def calculate_price(self): # rule is class ka lia hia 
        pass

# child class
class AC(Product):
    def calculate_price(self):
        return self.get_price()

# child class
class Refrigerator(Product):
    def calculate_price(self):
        return self.get_price()

# child class
class WashingMachine(Product):
    def calculate_price(self):
        return self.get_price()

# main custoemr 
class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, product):
        if product.stock > 0:
            self.cart.append(product)
            product.stock -= 1
        else:
            print(f"""{product.name} is out of stock """)

    def calculate_bill(self):
        total = sum(p.calculate_price() for p in self.cart)

        # Discount
        discount = 0
        if total > 100000:
            discount = total * 0.10

        after_discount = total - discount

        # Tax 18%
        tax = after_discount * 0.18

        final_amount = after_discount + tax

        return total, discount, tax, final_amount

    # Invoice
    def generate_invoice(self):
        invoice_id = uuid.uuid4()

        total, discount, tax, final = self.calculate_bill()

        invoice = (
            f"Invoice ID: {invoice_id}\n"
            f"Customer: {self.name}\n"
            f"Total: {total}\n"
            f"Discount: {discount}\n"
            f"Tax (18%): {tax}\n"
            f"Final Amount: {final}\n"
        
        )

        with open("invoice.txt", "a") as f:
            f.write(invoice)

        return invoice
```

```python
ac = AC("Gree AC", 120000, 5)
ref = Refrigerator("Haier Fridge", 90000, 3)
wm = WashingMachine("LG Washer", 70000, 2)
```

```python
customer = Customer("Hamza")

customer.add_to_cart(ac)
customer.add_to_cart(ref)
customer.add_to_cart(wm)
```

```python
print(customer.generate_invoice())
```
